```python
import pandas as pd
beta_df = pd.read_csv('./results/multidisorder_median_betas_parc.csv')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the figure size for better readability
plt.figure(figsize=(10, 6))

# sns.stripplot automatically plots columns on X and their values on Y
sns.stripplot(data=beta_df.iloc[:, 1:], jitter=0.2, alpha=0.6, size=5)
plt.xlabel("Disorders")
plt.ylabel("LME beta")
plt.tight_layout()
plt.show()
```
